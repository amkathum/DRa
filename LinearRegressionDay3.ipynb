# Read data

```python
data = pd.read_csv("/content/Student_Performance.csv")
```

```python
data.head()
```

```python
# change category to numbers
data["Extracurricular Activities"] = data["Extracurricular Activities"].astype("category").cat.codes
```

```python
data.head()
```

```python
data.size
```

```python
X = data.drop("Performance Index", axis = 1) # axis = 1 for column , axis = 0 for rows
y = data["Performance Index"]
```

```python
X.head() # X1 to X5
```

```python
y.head() # label
```

```python
# for the bias or theta 0
ones = np.ones(X.shape[0]) # shape[0] => number of rows shape[1] => number of columns
```

```python
# numpy must have numpy arrays inside a tuple (nparr1,nparr2)
X = np.column_stack((ones,X.to_numpy()))
```

```python
X
```

```python
# Split
from sklearn.model_selection import train_test_split
# split data 80% 20% => 80% for train , 20% for test

# X, test_size = 0.2 aka 20% => 0.8 * X , 0.2 * X
# if X had 10,000 => 8000 , 2000
# if y had 10,000 => 8000 , 2000
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size = 0.2)
```

# Find theta

```python
# only use training !
theta = np.linalg.solve(X_train.T @ X_train, X_train.T @ y_train)
```

```python
theta
```

```python
# dot product
predictions = X_test @ theta
```

```python
features = np.array([1,9,100,0,10,1])
```

```python
# y = C+ D t => y = theta + theta * X + theta * X
# y = m*x+b
features.T @ theta
```

```python
predictions
```

```python
# mean square erro
def mean_square_error(y_true,y_pred):

  error = y_true - y_pred

  squared_error = error **2

  mse = squared_error.mean()

  return mse
```

```python
'''
  if mse for train and test near each other it is good
  if not, might be overfitting
'''
mean_square_error(predictions,y_test)
```

```python
mean_square_error(X_train @ theta, y_train)
```

# Sklearn

```python
# import
from sklearn.linear_model import LinearRegression
```

```python
# train
reg = LinearRegression().fit(X_train,y_train)
# predict
reg.predict(X_test)
```

```python
reg.coef_
```

```python
print(reg.score(X_train,y_train))
```

```python
print(reg.score(X_test,y_test))
```

```python
mean_square_error(reg.predict(X_test),y_test)
```

```python
mean_square_error(reg.predict(X_train),y_train)
```
